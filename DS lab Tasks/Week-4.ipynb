```python
import matplotlib.pyplot as plt
y1=[] #stores +ve values
y2=[] #stores -ve values
x = range(-100, 100, 10) #Generate values form -100 to 90
for i in x: y1.append(i**2)
for i in x: y2.append(-i**2)

plt.plot(x, y1)
plt.plot(x, y2)
plt.xlabel("x")
plt.xlabel("y")
plt.ylim(-2000, 2000)
plt.axhline(0)
plt.axvline(0)
plt.savefig("quad.png")
plt.show()
```

```python
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=['sepal length (cm)',
             'sepal width (cm)',
             'petal length (cm)',
             'petal width (cm)']
)

print(df.head())
plt.subplot(2, 3, 1)
plt.plot(df['sepal length (cm)'], df[('petal length (cm)')])

plt.subplot(2, 3, 2)
plt.plot(df['sepal length (cm)'], df[('petal length (cm)')])

plt.subplot(2, 3, 2)
plt.plot(df['sepal length (cm)'], df[('petal length (cm)')], c ='green')

plt.title('subplots')
plt.show()
```

```python
from numpy import np
import seaborn as sns
import matplotlib.pyplot as plt
iris=sns.load_dataset("iris")
setosa = iris.loc[iris.species == "setosa"]
sns.kdeplot(
    x=setosa.petal_length,
    y=setosa.petal_width,
    fill=True
)
plt.title("KDE plot of setosa Petal Dimension")
plt.xlabel("Petal length")
plt.ylabel("petal width")
plt.show()
```
